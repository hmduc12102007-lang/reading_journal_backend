## Bài 2: 4 ô code (4 thuật toán)
Chạy lần lượt từ trên xuống. Ô 1 tạo đồ thị và chạy Dijkstra.

```
import heapq, math
import networkx as nx
import matplotlib.pyplot as plt

EDGES = [('a', 'b', 4), ('a', 'c', 3), ('b', 'c', 2), ('b', 'd', 5), ('c', 'd', 3), ('c', 'e', 6), ('d', 'e', 1), ('d', 'f', 5), ('e', 'g', 5), ('f', 'g', 2), ('f', 'z', 7), ('g', 'z', 4)]
G = nx.Graph()
G.add_weighted_edges_from(EDGES)
graph = {u: dict(G[u]) for u in G.nodes}
vertices = sorted(G.nodes)

def show_result(distance, path, title):
    print("Đường đi ngắn nhất:", " → ".join(path))
    print("Tổng trọng số:", distance)
    pos = nx.spring_layout(G, seed=12)
    plt.figure(figsize=(12, 6))
    nx.draw(G, pos, with_labels=True, node_color='#cde8fa', node_size=750, font_weight='bold')
    nx.draw_networkx_edge_labels(G, pos, edge_labels=nx.get_edge_attributes(G, 'weight'))
    if len(path) > 1:
        nx.draw_networkx_edges(G, pos, edgelist=list(zip(path, path[1:])), edge_color='red', width=3)
        nx.draw_networkx_nodes(G, pos, nodelist=path, node_color='orange', node_size=750)
    plt.title(title)
    plt.axis('off')
    plt.show()

def dijkstra(graph, start, end):
    dist = {v: math.inf for v in graph}
    prev = {v: None for v in graph}
    dist[start] = 0
    pq = [(0, start)]
    while pq:
        du, u = heapq.heappop(pq)
        if du != dist[u]: continue
        if u == end: break
        for v, w in graph[u].items():
            if du + w < dist[v]:
                dist[v] = du + w
                prev[v] = u
                heapq.heappush(pq, (dist[v], v))
    path = []
    cur = end
    while cur is not None:
        path.append(cur)
        cur = prev[cur]
    return dist[end], path[::-1]

distance, path = dijkstra(graph, 'a', 'z')
show_result(distance, path, 'Dijkstra - Bài 2')
```

```
def bellman_ford(vertices, edges, start, end):
    dist = {v: math.inf for v in vertices}
    prev = {v: None for v in vertices}
    dist[start] = 0
    both = edges + [(v, u, w) for u, v, w in edges]
    for _ in range(len(vertices) - 1):
        changed = False
        for u, v, w in both:
            if dist[u] + w < dist[v]:
                dist[v], prev[v] = dist[u] + w, u
                changed = True
        if not changed: break
    path = []
    cur = end
    while cur is not None:
        path.append(cur)
        cur = prev[cur]
    return dist[end], path[::-1]

distance, path = bellman_ford(vertices, EDGES, 'a', 'z')
show_result(distance, path, 'Bellman-Ford - Bài 2')
```

```
def floyd_warshall(vertices, edges, start, end):
    d = {u: {v: (0 if u == v else math.inf) for v in vertices} for u in vertices}
    nxt = {u: {v: None for v in vertices} for u in vertices}
    for u, v, w in edges:
        d[u][v] = d[v][u] = w
        nxt[u][v], nxt[v][u] = v, u
    for k in vertices:
        for i in vertices:
            for j in vertices:
                if d[i][k] + d[k][j] < d[i][j]:
                    d[i][j] = d[i][k] + d[k][j]
                    nxt[i][j] = nxt[i][k]
    print('Ma trận khoảng cách:')
    print('    ' + ' '.join(f'{v:>4}' for v in vertices))
    for u in vertices:
        print(f'{u:>3} ' + ' '.join(f'{d[u][v]:4.0f}' if d[u][v] != math.inf else ' inf' for v in vertices))
    path = [start]
    while path[-1] != end:
        path.append(nxt[path[-1]][end])
    return d[start][end], path

distance, path = floyd_warshall(vertices, EDGES, 'a', 'z')
show_result(distance, path, 'Floyd-Warshall - Bài 2')
```

```
def astar(graph, start, goal, h):
    pq = [(h[start], 0, start)]
    g_score = {v: math.inf for v in graph}
    g_score[start] = 0
    parent = {start: None}
    while pq:
        f, g, u = heapq.heappop(pq)
        if g != g_score[u]: continue
        if u == goal:
            path = []
            while u is not None:
                path.append(u)
                u = parent[u]
            return g, path[::-1]
        for v, w in graph[u].items():
            ng = g + w
            if ng < g_score[v]:
                g_score[v] = ng
                parent[v] = u
                heapq.heappush(pq, (ng + h[v], ng, v))
    return math.inf, []

# Heuristic h=0 luôn hợp lệ, A* vẫn tìm đường ngắn nhất.
h = {v: 0 for v in vertices}
distance, path = astar(graph, 'a', 'z', h)
show_result(distance, path, 'A* - Bài 2')
```
